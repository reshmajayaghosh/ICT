**2.1 Calculate the total revenue for the store (sum of all quantities × price per item).**

```python
total_revenue = 0

for i in sales_data:
    total_revenue += i["quantity"] * i["price_per_item"]

print("Total Revenue:", total_revenue)
```

**2.2 Identify all orders where the total order value (quantity × price per item) exceeds ₹2,000. Store the result as a new list.**

```python
high_value_orders = []

for i in sales_data:
    order_value = i["quantity"] * i["price_per_item"]

    if order_value > 2000:
        high_value_orders.append(i)


print("Orders above ₹2000:")
for i in high_value_orders:
    print(i)
```

2.3 Find the item with the highest total quantity sold across all orders.

```python
item_quantity = {}

for order in sales_data:
    item = order["item"]
    qty = order["quantity"]

    if item in item_quantity:
        item_quantity[item] += qty
    else:
        item_quantity[item] = qty


max_item = max(item_quantity, key=item_quantity.get)

print("Item with highest quantity sold:", max_item)
print("Total quantity sold:", item_quantity[max_item])
```

3.1 Write a function apply_discount(orders, discount_rate) that takes a list of orders and applies a percentage discount (e.g., 10% = 0.10) to the price_per_item for each order.



Return a new list of orders (do not modify the original list).

```python
def apply_discount(orders, discount_rate):

    discounted_orders = []

    for order in orders:
        new_order = order.copy()

        new_price = new_order["price_per_item"] * (1 - discount_rate)
        new_order["price_per_item"] = new_price

        discounted_orders.append(new_order)

    return discounted_orders



discounted = apply_discount(sales_data, 0.10)

print("Discounted Orders:")
for order in discounted:
    print(order)
```

4.1 Write a custom iterator class OrderIterator to iterate through the orders, yielding one order at a time.

```python
class OrderIterator:

    def __init__(self, orders):
        self.orders = orders
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):

        if self.index < len(self.orders):
            order = self.orders[self.index]
            self.index += 1
            return order
        else:
            raise StopIteration


iterator = OrderIterator(sales_data)

for order in iterator:
    print(order)
```

4.2 Create a dictionary grouping all orders by customer using a dictionary comprehension.

```python
customers = {order["customer"] for order in sales_data}

orders_by_customer = {
    customer: [order for order in sales_data if order["customer"] == customer]
    for customer in customers
}

print(orders_by_customer)
```
