# 3. Outlier Detection and Handling

```python
for i in df.columns:
  plt.figure(figsize = (10,5))
  sns.boxplot(df[i])
  plt.show()
```

*Removing all the outliers from the numerical columns except the target column*

```python
num_cols = df.select_dtypes('number').columns.drop('Sale Price')
num_cols
```

```python
def handle_outliers(df, columns):
  for i in columns:
    if df[i].dtype in ['int64', 'float64']:
      q1 = df[i].quantile(0.25)
      q3 = df[i].quantile(0.75)
      iqr = q3 - q1
      lower_bound = q1 - 1.5 * iqr
      upper_bound = q3 + 1.5 * iqr
      df[i] = df[i].clip(lower_bound,upper_bound)

  return df
```

```python
df[num_cols] = handle_outliers(df[num_cols],df[num_cols].columns)
```

# `Step 3 : Model Training, Evaluationa and Selection`

Training the model with test size = 0.2

# Data Splitting

```python
x = df.drop(columns = 'Sale Price')
y = df['Sale Price']
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
x_train.shape
```

```python
y_test.shape
```

# Training the model

**Linear Regression**

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

r2_score(y_test,y_pred)
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
mean_squared_error(y_test,y_pred)
```

*R2 Score is not that and there this is not performing well. Model is not that great*

**Applying Polynomial Regression with degree 2**

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree = 2)
x_train_poly = poly.fit_transform(x_train)
x_test_poly = poly.transform(x_test)
```

```python
model = LinearRegression()
model.fit(x_train_poly,y_train)
y_pred = model.predict(x_test_poly)
```

```python
r2_score(y_test,y_pred)
```
